```python
import matplotlib.pyplot as plt
import numpy as np

cm_spo2_ecg_Ftest = np.array([[1384, 515, 183], 
                               [465, 988, 632], 
                               [254, 551, 1280]])

cm_spo2_Ftest = np.array([[1367, 513, 202],
                          [427, 944, 714], 
                          [246, 512, 1327]])

cm_ecg_Ftest = np.array([[977, 390, 715],
                         [781, 422, 882],
                         [537, 334, 1214]])

labels = ['A', 'H', 'N']
titles = ['SpO2 + ECG', 'SpO2', 'ECG']
letters = ['(A)', '(B)', '(C)']

def plot_confusion_matrix(ax, cm, label_names, title, letter, idx):
    
    im = ax.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    fig = ax.get_figure()
    #fig.colorbar(im, ax=ax)
    
    tick_marks = np.arange(len(label_names))
    ax.set_xticks(tick_marks)
    ax.set_yticks(tick_marks)
    ax.set_xticklabels(label_names)
    ax.set_yticklabels(label_names)
    ax.tick_params(axis='both', which='major', labelsize=14)
    
    if idx == 1:
        ax.set_xlabel('PREDICTED LABEL', fontsize=14, fontweight='bold')

    if idx == 0:
        ax.set_ylabel('ACTUAL LABEL', fontsize=14, fontweight='bold')
    ax.set_title(f"{letter} {title}", fontsize=18, fontweight='bold', pad=15)

    # add the numbers inside the squares
    thresh = cm.max() / 2. 
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, format(cm[i, j], 'd'),
                    ha="center", va="center", fontsize=15,
                    color="white" if cm[i, j] > thresh else "black")


fig, axes = plt.subplots(1, 3, figsize=(12, 4))

data_list = [cm_spo2_ecg_Ftest, cm_spo2_Ftest, cm_ecg_Ftest]

for i in range(3):
    plot_confusion_matrix(axes[i], data_list[i], labels, titles[i], letters[i], i)

plt.tight_layout()
plt.show()
```
